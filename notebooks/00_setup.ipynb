# Setup
Creates the schema and the volume used by the pipeline. Run it once, before the other notebooks.

After running it, upload the input file through **Catalog → workspace → retail → Volumes → raw → Upload to this volume**:

| Volume | File | Source |
|---|---|---|
| `raw` | `online_retail_II.xlsx` | [UCI Machine Learning Repository: Online Retail II](https://archive.ics.uci.edu/dataset/502/online+retail+ii) (CC BY 4.0) |

```sql
-- All objects are idempotent: the notebook can be re-run safely
CREATE SCHEMA IF NOT EXISTS workspace.retail
COMMENT 'UCI Online Retail II transactions (UK online gift retailer, Dec 2009 - Dec 2011). Medallion pipeline and customer churn model.';

CREATE VOLUME IF NOT EXISTS workspace.retail.raw
COMMENT 'Raw input file (online_retail_II.xlsx), as downloaded from the UCI repository';
```
